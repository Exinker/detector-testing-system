```python
import os

import numpy as np

from detector_testing_system.characteristic.current import (
    BaseCurrentModel,
    JNormCurrentModel,
    calculate_current,
    research_current,
)
from detector_testing_system.characteristic.gradient import calculate_gradient
from detector_testing_system.data import filter_trace_factory, load_data
from detector_testing_system.device import run_device
from detector_testing_system.experiment import ExperimentConfig, run_experiment
```

```python
LABEL = 'MP14E015_960ns_4us_exp89_t20_ABD=12.4_DK'
THRESHOLD = (0, 40)
INTERVAL = (0, +np.inf)
```

```python
config = ExperimentConfig.from_ini(LABEL)
device = run_device(
    config=config,
)

run_experiment(
    device=device,
    config=config,
    params=[
        (100, np.arange(2, 50+2, 2)),
    ],
    label=os.path.join(LABEL, 'dark-current'),
    force=True,
)
```

```python
data = load_data(
    label=os.path.join(LABEL, 'dark-current'),
    show=True,
)
```

```python
current = calculate_current(
    trace=data.trace(42),
    model=BaseCurrentModel(
        weighted=True,
        filter=filter_trace_factory(
            threshold=THRESHOLD,
            interval=INTERVAL,
        ),
    ),
)
current.show()
```

```python
current = calculate_current(
    trace=data.trace(42),
    model=JNormCurrentModel(),
)
current.show()
```

```python
current = research_current(
    data=data,
    model=BaseCurrentModel(),
)
current.show(
    note='demo',  # добавить заметку
)
```

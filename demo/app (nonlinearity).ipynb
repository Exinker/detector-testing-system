```python
import os

import numpy as np

from detector_testing_system.characteristic.gradient import calculate_gradient
from detector_testing_system.characteristic.nonlinearity import (
    BaseDarkCurrentModel,
    JNormDarkCurrentModel,
    calculate_nonlinearity,
    compare_nonlinearity,
    research_nonlinearity,
)
from detector_testing_system.data import filter_trace_factory, load_data
from detector_testing_system.device import run_device
from detector_testing_system.experiment import ExperimentConfig, run_experiment
```

```python
LABEL = 'MP14E015_960ns_4us_exp89_t20_ABD=12.4_DK'
THRESHOLD = (0, 40)
INTERVAL = (0, +np.inf)
```

```python
config = ExperimentConfig.from_ini(LABEL)
device = run_device(
    config=config,
)

run_experiment(
    device=device,
    config=config,
    params=[
        (100, np.arange(2, 50+2, 2)),
    ],
    label=os.path.join(LABEL, 'nonlinearity'),
    force=True,
)
```

```python
data = load_data(
    label=os.path.join(LABEL, 'nonlinearity'),
    show=True,
)
```

```python
gradient = calculate_gradient(
    trace=data.trace(42),
)
gradient.show()
```

```python
nonlinearity = calculate_nonlinearity(
    trace=data.trace(42),
    model=BaseDarkCurrentModel(
        weighted=True,
        filter=filter_trace_factory(
            threshold=THRESHOLD,
            interval=INTERVAL,
        ),
    ),
    verbose=False,  # отключить отображение способа расчета
)
nonlinearity.show(
    note='demo',  # добавить заметку
)
```

```python
nonlinearity = calculate_nonlinearity(
    trace=data.trace(42),
    model=JNormDarkCurrentModel(),
    verbose=False,  # отключить отображение способа расчета
)
nonlinearity.show(
    views=[
        None,
        {'ylim': [-50, +50]},  # установить xlim для правого графика
    ],
    note='demo',  # добавить заметку
)
```

```python
nonlinearity = research_nonlinearity(
    data=data,
    model=BaseDarkCurrentModel(),
)
nonlinearity.show(
    note='demo',  # добавить заметку
)
```

```python
nonlinearity = research_nonlinearity(
    data=data,
    model=JNormDarkCurrentModel(),
)
nonlinearity.show(
    note='demo',  # добавить заметку
)
```

```python
compare_nonlinearity(
    [
        (data.trace(42-1), r'$U_{\alpha-1}$'),
        (data.trace(42+1), r'$U_{\alpha+1}$'),
    ],
    model=JNormDarkCurrentModel(),
    views=[
        None,
        {'ylim': [-50, +50]},   # установить xlim для правого графика
    ],
)
```
